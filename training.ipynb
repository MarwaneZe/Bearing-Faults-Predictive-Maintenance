```python
def test_with_condition(metadata):
    """Return a copy with a 'condition' column like 'N15_M07_F10'."""
    df = metadata.copy()
    df["condition"] = (
        "N" + (df["speed_rpm"] // 100).astype(str).str.zfill(2)
        + "_M" + (df["torque_Nm"] * 10).round().astype(int).astype(str).str.zfill(2)
        + "_F" + (df["force_N"] // 100).astype(str).str.zfill(2)
    )
    return df
```
