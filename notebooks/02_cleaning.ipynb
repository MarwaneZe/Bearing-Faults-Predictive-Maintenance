# Cleaning — windowed storage, split-tagged (Option B)

**Input:** `data/processed/metadata.csv` (2557 usable recordings).

**Output:** `data/processed/window_index.csv` — one row per recording with the
number of windows it yields and its bearing-wise Split A fold + Split B
partition.


```python
# Build the split-tagged window index
from utils import build_metadata
from pipeline.window_index import build_window_index

metadata = build_metadata(refresh=False)
index = build_window_index(metadata, save=True) 
display(index.head())
```

```python
# Corpus + split summary
print(f"recordings: {len(index)} | total windows: {index['n_windows'].sum()}")
print("\nSplit A — bearings per fold (label):")
display(index.groupby(["fold", "label"])["bearing"].nunique().unstack(fill_value=0))
print("Split B — windows per partition:")
display(index.groupby("partition")["n_windows"].sum())
```

## Findings
- 2,557 recordings → **76,868 windows** (variable lengths absorbed by Option B).
- Split A: every fold holds ≥1 healthy + 5–6 damaged bearings.
- Split B: train = artificial (KA) + healthy; test = real/combination (KI/KB).
  The six healthy bearings act as the shared reference during evaluation.
